```python
import torch 
import torch.nn as nn

import functools
import os
import sys 

module_path = os.path.abspath(os.path.join('..'))
if module_path not in sys.path:
    sys.path.append(module_path)

from models import networks

def instance_norm(dimension, affine = False):
    if dimension == 2:
        return nn.InstanceNorm2d

    elif dimension == 3:
        return nn.InstanceNorm3d

    else:
        raise Exception('Invalid image dimension.')
    
norm_layer = functools.partial(instance_norm(3), affine=False, track_running_stats=True)

model = networks.UNet3DAnisotropic(1, 1,norm_layer)
```

```python
x = torch.randn((1, 1, 128, 128, 10))  # Batch of 1, grayscale, 3D image
output = model(x)
print(output.shape)  # Should be torch.Size([1, 1, 256, 256, 24])
```

```python
model_2 = networks.Unet_deconv(1, 1,norm_layer)
```

```python
x = torch.randn((1, 1, 128, 128, 128))  # Batch of 1, grayscale, 3D image
output = model_2(x)
print(output.shape)  # Should be torch.Size([1, 1, 256, 256, 24])
```

```python
x = torch.randn((1, 1, 128, 128, 20))  # Batch of 1, grayscale, 3D image
output = model_2(x)
print(output.shape)  # Should be torch.Size([1, 1, 256, 256, 24])
```

```python
norm_layer_2 = functools.partial(instance_norm(2), affine=False, track_running_stats=True)

model_3 = networks.NLayerDiscriminator(input_nc=1, ndf=64, n_layers=3, norm_layer=norm_layer_2)
```
